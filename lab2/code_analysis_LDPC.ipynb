# Analysis of the LDPC library
We are going to use the wrapper class to analyze different size of puncturing and shortening

```python
import numpy as np
```

```python
import data as data
import wrapper_class as ldpc
import libraries.file_utils as file_utils
import data_saving as data_saving
```

```python
computation = True #set to false to skip the computation and load previous results
```

# codes loading

```python
codes_size = data.code_size
code_file = data.codes_dict[codes_size]["file"]
n = data.codes_dict[codes_size]["n"]
```

```python
codes = file_utils.codes_from_file(code_file)
R_range = []
for code in codes:
    R_range.append(code[0])
print(f"R range: {R_range}")
```

# Manually rate vs. Auto rate

```python
qber_step = data.qber_step * 4
qber_list = np.arange(data.qber_start, data.qber_end + qber_step, qber_step)
print("QBERs to be tested:", qber_list)
```

### bsc channel

```python
save_file = data.saving_folder + "LDPC_bsc_auto.csv"
results_bsc_auto = []
print(f"save_file: {save_file}")
```

```python
if computation:
    for qber in qber_list:
        print(f"Testing QBER = {qber:.3f}")
        result = ldpc.test_ec_fixed(
            qber,
            R_range=R_range,
            codes=codes,
            n=n,
            n_tries=data.n_tries,
            channel=ldpc.bsc_channel,
            show=1,
            discl_k=1
        )
        data_saving.save_result_csv(result, save_file)
```

```python
results_bsc_auto = data_saving.load_result_csv(save_file)
print("Results loaded:")
print(results_bsc_auto)
```

### fixed error length channel

```python
save_file = data.saving_folder + "LDPC_fixed_auto.csv"
results_fixed_auto = []
print(f"save_file: {save_file}")
```

```python
if computation:
    for qber in qber_list:
        print(f"Testing QBER = {qber:.3f}")
        result = ldpc.test_ec_fixed(
            qber,
            R_range=R_range,
            codes=codes,
            n=n,
            n_tries=data.n_tries,
            channel=ldpc.fixed_error_channel,
            show=1,
            discl_k=1
        )
        data_saving.save_result_csv(result, save_file)
```

```python
results_fixed_auto = data_saving.load_result_csv(save_file)
print("Results loaded:")
print(results_fixed_auto)
```
