# Analysis of CASCADE protocol for different qber

```python
import numpy as np
```

```python
import data as data
import cascade as cascade
import data_saving as data_saving
```

```python
#Data parameter
qber_start=data.qber_start
qber_end=data.qber_end
qber_step=data.qber_step*4
rounds = 7
initial_constant = 0.73
```

```python
for qber in np.arange(qber_start, qber_end + qber_step, qber_step):
    print(f"Testing CASCADE at QBER = {qber:.3f}")
    block_sizes = cascade.generate_block_sizes(initial_constant, qber, rounds)
    print(f"Block sizes: {block_sizes}")
```

# chosing block lenght->equals to LDPC for comparison and applying CASCADE

```python
code_size = data.code_size
print(f"Code size: {code_size}")
```

## Test for bsc channel

```python
saving_file = f"{data.saving_folder}cascade_bsc_{code_size}_results.csv"
tries = 100
```

```python
for qber in np.arange(qber_start, qber_end + qber_step, qber_step):
    print(f"Testing CASCADE at QBER = {qber:.3f}")
    block_sizes = cascade.generate_block_sizes(initial_constant, qber, rounds)
    print(f"Block sizes: {block_sizes}")
    cascade_results = []
    for _ in range(tries):
        cascade_results.append(cascade.cascade_protocol(key_length=int(code_size), 
                                                channel_fn=cascade.bsc_channel,
                                                qber=qber, 
                                                block_sizes=block_sizes, 
                                                rounds=rounds))
    metrics = cascade.compute_metrics(cascade_results)
    print(f"metrics: {metrics}")
    data_saving.save_result_csv(metrics, filename=saving_file)
    print("---------------------------------------------------")
```

### Test for fixed channel

```python
saving_file = f"{data.saving_folder}cascade_fixed_{code_size}_results.csv"
```

```python
for qber in np.arange(qber_start, qber_end + qber_step, qber_step):
    print(f"Testing CASCADE at QBER = {qber:.3f}")
    block_sizes = cascade.generate_block_sizes(initial_constant, qber, rounds)
    print(f"Block sizes: {block_sizes}")
    cascade_results = []
    for _ in range(tries):
        cascade_results.append(cascade.cascade_protocol(key_length=int(code_size), 
                                                channel_fn=cascade.fixed_error_channel,
                                                qber=qber, 
                                                block_sizes=block_sizes, 
                                                rounds=rounds))
    metrics = cascade.compute_metrics(cascade_results)
    print(f"metrics: {metrics}")
    data_saving.save_result_csv(metrics, filename=saving_file)
    print("---------------------------------------------------")
```
